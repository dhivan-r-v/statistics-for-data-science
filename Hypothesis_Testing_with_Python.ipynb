## 1.	The historical average score in a certification exam is 65. A training institute introduces a new course. A random sample of 50 students trained under the new course has a mean score of 68 and a standard deviation of 10. At the 5% level of significance, test whether the new course improves performance.

1) Ho(Null Hypothesis) : There is no significant difference in the average score
2) Ha(Alternative Hypothesis) : There is a significant difference in the average score

```python
import numpy as np
from scipy import stats

sample_50=np.random.normal(loc=68, scale=5.0, size=50).astype("int")
t_stats,p_value=stats.ttest_1samp(a=sample_50,popmean=65,alternative='greater')

if p_value<0.05:
    print("We Reject H0")
else:
    print("We do not reject H0")
```

### At 5% level of significance,we can reject the null hypothesis and claim alternative hypothesis


## 2.	A company claims that the average time taken to resolve customer tickets is 30 minutes. A random sample of 40 tickets shows a mean resolution time of 33 minutes with a standard deviation of 8 minutes. At the 5% level of significance, test whether the average resolution time has increased.

1) Ho(Null Hypothesis) : There is no significant difference in the average time taken.
2) Ha(Alternative Hypothesis) : There is a significant difference in the average time taken.

```python
import numpy as np
from scipy import stats

sample_40 = np.random.normal(loc=33, scale=8.0, size=40)

t_stats, p_value = stats.ttest_1samp(
    a=sample_40,
    popmean=30,
    alternative='greater'
)

if p_value < 0.05:
    print("We Reject H0")
else:
    print("We do not reject H0")
```

### At 5% level of significance,we can reject the null hypothesis and claim alternative hypothesis

## 3.	A company claims that 60% of its users are satisfied with its product. In a survey of 200 users, 134 users report satisfaction. At the 5% level of significance, test whether customer satisfaction is higher than claimed.

1) Ho(Null Hypothesis) : There is no significant difference in customer satisfaction.
2) Ha(Alternative Hypothesis) : There is a significant difference in customer satisfaction.

```python
from statsmodels.stats.proportion import proportions_ztest

count = 134      # satisfied users
nobs = 200       # total users

z_stat, p_value = proportions_ztest(
    count=count,
    nobs=nobs,
    value=0.60,
    alternative='larger'
)

print("Z-statistic:", z_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("We Reject H0")
else:
    print("We do not reject H0")
```

### At 5% significance level we can reject null hypothesis and we can claim alternative hypothesis(customer satisfaction is higher thancompany's claimed rate of 60%)

## 5.	Two teaching methods are used to train employees. A sample of 30 employees trained using Method A has a mean score of 75 with a standard deviation of 12. A sample of 28 employees trained using Method B has a mean score of 70 with a standard deviation of 10. At the 5% level of significance, test whether there is a difference in average scores between the two methods.

1) Ho(Null Hypothesis) : There is no significant difference in the average score between the two tests.
2) Ha(Alternative Hypothesis) : There is a significant difference in the average score between the two tests

```python
import numpy as np
from scipy import stats

np.random.seed(42)

method_A = np.random.normal(loc=75, scale=12, size=30)
method_B = np.random.normal(loc=70, scale=10, size=28)

t_stat, p_value = stats.ttest_ind(
    method_A,
    method_B,
    equal_var=False
)

print(t_stat)
print(p_value)
if p_value < 0.05:
    print("We Reject H0")
else:
    print("We do not reject H0")
```

### At 5% significance level we cannot reject null hypothesis ,here is insufficient evidence to conclude that the average scores differ between employees trained using Method A and Method B. Although Method A has a higher observed mean score, the difference is not statistically significant.

## 6.	The average delivery time from two warehouses is compared using random samples. Warehouse A has a sample of 35 deliveries with a mean time of 48 minutes and a standard deviation of 6 minutes. Warehouse B has a sample of 40 deliveries with a mean time of 52 minutes and a standard deviation of 7 minutes. At the 5% level of significance, test whether the mean delivery times differ.

1) Ho(Null Hypothesis) : There is no significant difference in the average time taken between the two warehouses.
2) Ha(Alternative Hypothesis) : There is a significant difference in the average time taken between the two warehouses.


```python
import numpy as np
from scipy import stats

np.random.seed(42)
warehouse_A = np.random.normal(loc=48, scale=6, size=35)

warehouse_B = np.random.normal(loc=52, scale=7, size=40)

t_stat, p_value = stats.ttest_ind(
    warehouse_A,
    warehouse_B,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("We Reject H0")
else:
    print("We do not reject H0")
```

### At the 5% significance level we reject null hypothesis, there is sufficient evidence to conclude that the average delivery times of Warehouse A and Warehouse B are significantly different.

## 7.	Two versions of a website are tested. Out of 500 users for Version A, 320 users click the signup button. Out of 480 users for Version B, 290 users click the signup button. At the 5% level of significance, test whether the conversion rates differ.

1) Ho(Null Hypothesis) : There is no significant difference in conversion rate between the two websites.
2) Ha(Alternative Hypothesis) : There is a significant difference in  conversion rate between the two websites.

```python
from statsmodels.stats.proportion import proportions_ztest

count = [320, 290]
nobs = [500, 480]

z_stat, p_value = proportions_ztest(
    count=count,
    nobs=nobs,
    alternative='two-sided'
)

print("Z-statistic:", z_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("We Reject H0")
else:
    print("We do not reject H0")
```

### At the 5% significance level we cannot reject null hypothesis, there is insufficient evidence to conclude that the conversion rates of Website Version A and Website Version B are different.

# Anova Test

Why do we use stats.f_oneway()?

**stats.f_oneway() performs a One-Way ANOVA. It is used when comparing the means of three or more independent groups. Instead of performing multiple t-tests, ANOVA uses an F-statistic to compare the variation between groups with the variation within groups and determines whether at least one group mean differs significantly from the others**.

## 10.	A company wants to test whether gender and product preference are independent. The survey results are as follows:
●	Male: 60 prefer Product A, 40 prefer Product B
●	Female: 50 prefer Product A, 50 prefer Product B

At the 5% level of significance, test for independence.

1. Ho(Null Hypothesis) : The gender and product preference are independent.
2. Ha(Alternative Hypothesis) : The gender and product preference are dependent. 


```python
import numpy as np
from scipy import stats

data = np.array([
    [60, 40],
    [50, 50]
])

chi2_stat, p_value, dof, expected = stats.chi2_contingency(data)

print("Chi-Square Statistic:", chi2_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("We Reject H0")
else:
    print("We do not reject H0")
```

### At the 5% significance level we do not reject null hypothesis, there is insufficient evidence to conclude that gender and product preference are associated. Therefore, gender and product preference can be considered independent based on the survey data.

## 11. 	A hospital studies whether smoking status is related to disease occurrence. The data collected is as follows:
* Smokers: 70 have the disease, 30 do not
* Non-smokers: 40 have the disease, 60 do not

At the 5% level of significance, test whether smoking status and disease occurrence are independent. 

1. Ho(Null Hypothesis) : The smoker and the disease are independent.
2. Ha(Alternative Hypothesis) : The smoker and the disease are dependent. 

```python
import numpy as np
from scipy import stats

data = np.array([
    [70, 30],
    [40, 60]
])

chi2_stat, p_value, dof, expected = stats.chi2_contingency(data)

print("Chi-Square Statistic:", chi2_stat)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)

if p_value < 0.05:
    print("We Reject H0")
else:
    print("We do not reject H0")
```

### At 5% level of significance we can reject null hypothesis and claim alternative hypothesis,that there is sufficient statistical evidence to conclude that smoking status and disease occurrence are not independent

## 12.	A dice manufacturer claims that a die is fair. The die is rolled 120 times, and the observed frequencies are:
* Face 1: 18
* Face 2: 22
* Face 3: 20
* Face 4: 25
* Face 5: 17
* Face 6: 18
At the 5% level of significance, test whether the die is fair.

1. Ho(Null Hypothesis) : The die is fair.
2. Ha(Alternative Hypothesis) : The die is unfair. 


```python
import numpy as np
from scipy import stats

observed = np.array([
    18,
    22,
    20,
    25,
    17,
    18
])

expected = np.array([
    20,
    20,
    20,
    20,
    20,
    20
])

chi2_stat, p_value = stats.chisquare(
    f_obs=observed,
    f_exp=expected
)

print("Chi-Square Statistic:", chi2_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("We Reject H0")
else:
    print("We do not reject H0")
```

### At the 5% significance level we cannot reject null hypothesis, there is insufficient evidence to conclude that the die is unfair. The observed frequencies are consistent with a fair die.
